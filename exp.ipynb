```python
# ================================================================
# GLOBAL FOOD DELIVERY MARKET ANALYSIS - 2026
# Data Analytics & Exploration Series (DAES)
# ================================================================

# 1. IMPORT LIBRARIES

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import zipfile
from google.colab import files

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

print("=" * 70)
print("GLOBAL FOOD DELIVERY MARKET ANALYSIS")
print("=" * 70)


# 2. LOAD DATASET (WITH FILE / ZIP UPLOADER FOR COLAB)

print("\nPlease upload your dataset file (.csv or .zip containing the csv):")
uploaded = files.upload()

uploaded_file_name = None

# Extract zip if uploaded, otherwise find csv
for filename in uploaded.keys():
    if filename.endswith('.zip'):
        print(f"Extracting {filename}...")
        with zipfile.ZipFile(filename, 'r') as zip_ref:
            zip_ref.extractall(".")
            # Search for extracted CSVs
            for extracted_file in zip_ref.namelist():
                if extracted_file.endswith('.csv'):
                    uploaded_file_name = extracted_file
                    break
        if uploaded_file_name:
            break
    elif filename.endswith('.csv'):
        uploaded_file_name = filename
        break

if not uploaded_file_name:
    raise FileNotFoundError(
        "No CSV dataset found in the uploaded file. Please run this cell again and upload your dataset."
    )

file_path = uploaded_file_name
df = pd.read_csv(file_path)

print("\nDATASET LOADED SUCCESSFULLY!")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nFIRST FIVE RECORDS:")
display(df.head())


# 3. DATASET INFORMATION

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nColumn Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all").T)


# 4. DATA CLEANING

print("\n" + "=" * 70)
print("DATA CLEANING")
print("=" * 70)

print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Records Before Cleaning:")
print(df.duplicated().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Remove completely empty columns
df = df.dropna(axis=1, how="all")

# Identify numerical and categorical columns
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

# Fill numerical missing values using median
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values
for col in categorical_columns:
    df[col] = df[col].fillna("Unknown")

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDataset Shape After Cleaning:", df.shape)


# 5. DATE AND TIME PREPROCESSING

print("\n" + "=" * 70)
print("DATE AND TIME PREPROCESSING")
print("=" * 70)

datetime_columns = []

for col in df.columns:
    if "date" in col.lower() or "timestamp" in col.lower():

        converted = pd.to_datetime(df[col], errors="coerce")

        if converted.notna().sum() > 0:
            df[col] = converted
            datetime_columns.append(col)

            # Create additional date features
            df[col + "_year"] = df[col].dt.year
            df[col + "_month"] = df[col].dt.month
            df[col + "_month_name"] = df[col].dt.month_name()
            df[col + "_day"] = df[col].dt.day

print("Detected Date Columns:", datetime_columns)


# 6. IDENTIFY DATA TYPES

numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

datetime_columns = df.select_dtypes(
    include=["datetime64[ns]"]
).columns.tolist()

print("\nNumerical Columns:")
print(numeric_columns)

print("\nCategorical Columns:")
print(categorical_columns)

print("\nDate Columns:")
print(datetime_columns)


# 7. AUTOMATIC COLUMN DETECTION

def find_column(keywords, columns):

    for keyword in keywords:
        for col in columns:

            if keyword in col.lower():
                return col

    return None


location_col = find_column(
    ["city", "location", "country", "region"],
    categorical_columns
)

food_col = find_column(
    ["food_category", "food_type", "cuisine", "category"],
    categorical_columns
)

payment_col = find_column(
    ["payment_method", "payment_type", "payment"],
    categorical_columns
)

restaurant_col = find_column(
    ["restaurant_name", "restaurant"],
    categorical_columns
)

rating_col = find_column(
    ["rating", "review_score"],
    numeric_columns
)

delivery_col = find_column(
    ["delivery_time", "delivery_duration"],
    numeric_columns
)

revenue_col = find_column(
    ["revenue", "order_value", "total_amount", "price"],
    numeric_columns
)

print("\nDETECTED IMPORTANT COLUMNS")

print("Location:", location_col)
print("Food Category:", food_col)
print("Payment Method:", payment_col)
print("Restaurant:", restaurant_col)
print("Customer Rating:", rating_col)
print("Delivery Time:", delivery_col)
print("Revenue / Order Value:", revenue_col)


# 8. EXPLORATORY DATA ANALYSIS

print("\n" + "=" * 70)
print("EXPLORATORY DATA ANALYSIS")
print("=" * 70)

print("\nNumerical Data Summary:")
display(df.select_dtypes(include=np.number).describe().T)

print("\nCategorical Data Summary:")
display(df.describe(include=["object", "category"]).T)


# 9. CREATE OUTPUT FOLDER

output_folder = "Food_Delivery_Analysis_Output"
graph_folder = os.path.join(output_folder, "Graphs")

os.makedirs(graph_folder, exist_ok=True)

graph_count = 0


# 10. GRAPH SAVING FUNCTION

def save_graph(filename):

    global graph_count

    plt.tight_layout()

    plt.savefig(
        os.path.join(graph_folder, filename),
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()

    graph_count += 1


# 11. VISUALIZATION 1 - LOCATION ANALYSIS

if location_col:

    plt.figure(figsize=(12, 6))

    location_data = df[location_col].value_counts().head(10)

    sns.barplot(
        x=location_data.values,
        y=location_data.index
    )

    plt.title("Top 10 Locations by Number of Records")
    plt.xlabel("Number of Records")
    plt.ylabel("Location")

    save_graph("01_Location_Analysis.png")


# 12. VISUALIZATION 2 - REVENUE DISTRIBUTION

if revenue_col:

    plt.figure(figsize=(10, 5))

    sns.histplot(
        df[revenue_col].dropna(),
        kde=True,
        bins=30
    )

    plt.title("Revenue / Order Value Distribution")
    plt.xlabel(revenue_col)
    plt.ylabel("Frequency")

    save_graph("02_Revenue_Distribution.png")


# 13. VISUALIZATION 3 - DELIVERY TIME

if delivery_col:

    plt.figure(figsize=(10, 5))

    sns.histplot(
        df[delivery_col].dropna(),
        kde=True,
        bins=30
    )

    plt.title("Food Delivery Time Distribution")
    plt.xlabel("Delivery Time")
    plt.ylabel("Frequency")

    save_graph("03_Delivery_Time.png")


# 14. VISUALIZATION 4 - CUSTOMER RATINGS

if rating_col:

    plt.figure(figsize=(10, 5))

    sns.histplot(
        df[rating_col].dropna(),
        kde=True,
        bins=20
    )

    plt.title("Customer Rating Distribution")
    plt.xlabel("Customer Rating")
    plt.ylabel("Frequency")

    save_graph("04_Customer_Ratings.png")


# 15. VISUALIZATION 5 - FOOD CATEGORY

if food_col:

    plt.figure(figsize=(12, 6))

    food_data = df[food_col].value_counts().head(10)

    sns.barplot(
        x=food_data.values,
        y=food_data.index
    )

    plt.title("Top 10 Food Categories")
    plt.xlabel("Number of Records")
    plt.ylabel("Food Category")

    save_graph("05_Food_Categories.png")


# 16. VISUALIZATION 6 - ORDER VALUE VS DELIVERY TIME

if revenue_col and delivery_col:

    plt.figure(figsize=(10, 6))

    sns.scatterplot(
        data=df,
        x=revenue_col,
        y=delivery_col,
        alpha=0.6
    )

    plt.title("Order Value vs Delivery Time")
    plt.xlabel("Order Value / Revenue")
    plt.ylabel("Delivery Time")

    save_graph("06_Order_vs_Delivery.png")


# 17. VISUALIZATION 7 - RESTAURANT ANALYSIS

if restaurant_col:

    plt.figure(figsize=(12, 6))

    restaurant_data = df[restaurant_col].value_counts().head(10)

    # Escaping dollar signs to avoid triggering Matplotlib LaTeX math mode parsing errors
    escaped_index = [str(x).replace("$", "\\$") for x in restaurant_data.index]

    sns.barplot(
        x=restaurant_data.values,
        y=escaped_index
    )

    plt.title("Top 10 Restaurants by Number of Records")
    plt.xlabel("Number of Records")
    plt.ylabel("Restaurant Price Tier")

    save_graph("07_Restaurant_Analysis.png")


# 18. VISUALIZATION 8 - MONTHLY TREND

if datetime_columns:

    date_col = datetime_columns[0]

    monthly_data = (
        df.set_index(date_col)
        .resample("ME")
        .size()
    )

    plt.figure(figsize=(12, 5))

    monthly_data.plot(marker="o")

    plt.title("Monthly Food Delivery Record Trend")
    plt.xlabel("Month")
    plt.ylabel("Number of Records")

    save_graph("08_Monthly_Trend.png")


# 19. VISUALIZATION 9 - PAYMENT METHODS

if payment_col:

    plt.figure(figsize=(10, 6))

    payment_data = df[payment_col].value_counts().head(10)

    plt.pie(
        payment_data.values,
        labels=payment_data.index,
        autopct="%1.1f%%",
        startangle=90
    )

    plt.title("Payment Method Distribution")

    save_graph("09_Payment_Methods.png")


# 20. VISUALIZATION 10 - CORRELATION HEATMAP

analysis_columns = [
    col for col in numeric_columns
    if not any(
        word in col.lower()
        for word in ["id", "code", "zip", "postal"]
    )
]

if len(analysis_columns) >= 2:

    plt.figure(figsize=(12, 8))

    correlation = df[analysis_columns].corr()

    sns.heatmap(
        correlation,
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Correlation Heatmap")

    save_graph("10_Correlation_Heatmap.png")


# 21. VISUALIZATION 11 - OUTLIER DETECTION

if numeric_columns:

    selected_columns = numeric_columns[:6]

    plt.figure(figsize=(12, 6))

    sns.boxplot(data=df[selected_columns])

    plt.title("Numerical Data - Outlier Detection")
    plt.xticks(rotation=45)

    save_graph("11_Outlier_Detection.png")


# 22. OUTLIER ANALYSIS USING IQR

print("\n" + "=" * 70)
print("OUTLIER ANALYSIS")
print("=" * 70)

outlier_summary = []

for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    outlier_summary.append({
        "Column": col,
        "Outlier Count": len(outliers),
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit
    })

outlier_df = pd.DataFrame(outlier_summary)

print("\nOutlier Summary:")
display(outlier_df)


# 23. CORRELATION ANALYSIS

print("\n" + "=" * 70)
print("CORRELATION ANALYSIS")
print("=" * 70)

if len(analysis_columns) >= 2:

    correlation = df[analysis_columns].corr()

    print("\nCorrelation Matrix:")
    display(correlation)

    corr_pairs = correlation.where(
        np.triu(
            np.ones(correlation.shape),
            k=1
        ).astype(bool)
    ).stack()

    if not corr_pairs.empty:

        print("\nTop Correlated Variable Pairs:")

        print(
            corr_pairs.abs()
            .sort_values(ascending=False)
            .head(5)
        )


# 24. BUSINESS INSIGHTS

print("\n" + "=" * 70)
print("KEY BUSINESS INSIGHTS")
print("=" * 70)

if location_col:

    print("\nTop Location:")

    print(df[location_col].value_counts().head(1))


if food_col:

    print("\nMost Popular Food Category:")

    print(df[food_col].value_counts().head(1))


if payment_col:

    print("\nMost Used Payment Method:")

    print(df[payment_col].value_counts().head(1))


if restaurant_col:

    print("\nRestaurant With Most Records:")

    print(df[restaurant_col].value_counts().head(1))


if rating_col:

    print("\nAverage Customer Rating:")

    print(round(df[rating_col].mean(), 2))


if delivery_col:

    print("\nAverage Delivery Time:")

    print(round(df[delivery_col].mean(), 2))


if revenue_col:

    print("\nAverage Revenue / Order Value:")

    print(round(df[revenue_col].mean(), 2))


print("\nTotal Records After Cleaning:", len(df))


# 25. SAVE CLEANED DATASET

cleaned_file = os.path.join(
    output_folder,
    "Cleaned_Global_Food_Delivery_Dataset.csv"
)

df.to_csv(cleaned_file, index=False)

print("\nCleaned Dataset Saved Successfully!")


# 26. SAVE OUTLIER REPORT

outlier_file = os.path.join(
    output_folder,
    "Outlier_Analysis_Report.csv"
)

outlier_df.to_csv(outlier_file, index=False)


# 27. SAVE ANALYSIS GRAPHS IN ZIP FILE

zip_filename = os.path.join(
    output_folder,
    "Food_Delivery_Analysis_Graphs.zip"
)

with zipfile.ZipFile(zip_filename, "w") as zipf:

    for root, dirs, files_list in os.walk(graph_folder):

        for filename in files_list:

            full_path = os.path.join(root, filename)

            zipf.write(
                full_path,
                arcname=filename
            )


# 28. FINAL CONCLUSION

print("\n" + "=" * 70)
print("PROJECT CONCLUSION")
print("=" * 70)

print("""
The Global Food Delivery Market Analysis project uses data analytics
to understand food delivery market trends and customer behaviour.

The dataset was cleaned by handling missing values, duplicate records
and empty columns.

Data preprocessing was performed to identify numerical, categorical
and date-related information.

Exploratory Data Analysis was conducted using statistical summaries
and graphical visualizations.

The project examines:

1. Location-wise record distribution.
2. Food category patterns.
3. Customer rating distribution.
4. Delivery time patterns.
5. Revenue and order value distribution.
6. Payment method preferences.
7. Restaurant record distribution.
8. Monthly record trends.
9. Correlations between numerical variables.
10. Possible outliers.

These findings can help food delivery businesses understand their
operations, identify patterns and make data-driven decisions.

Note: The conclusions are based on the uploaded dataset. Correlation
does not establish causation, and the results depend on the quality
and meaning of the dataset columns.
""")


# 29. FINAL OUTPUT

print("\n" + "=" * 70)
print("ANALYSIS COMPLETED SUCCESSFULLY!")
print("=" * 70)

print("\nTotal Graphs Generated:", graph_count)

print("\nOutput Folder:")
print(output_folder)

print("\nGenerated Files:")
print("1. Cleaned Dataset CSV")
print("2. Outlier Analysis Report CSV")
print("3. Graphs Folder")
print("4. Graphs ZIP File")

print("\nAll files are saved inside Food_Delivery_Analysis_Output directory for download.")
```
